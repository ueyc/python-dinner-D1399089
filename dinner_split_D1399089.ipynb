## 題目說明
將含有貨幣符號與逗號的餐費文字、文字型態的人數整理並轉換型別後，依規則計算：

1. 服務費 = 原始餐費 × 10%
2. 應付總額 = 原始餐費 + 服務費 − 折價券
3. 每人費用 = 應付總額 ÷ 分攤人數

最後以比較運算判斷每人費用是否超過每人預算 300 元，並輸出費用摘要。

## 測試一：原始資料（4 人、折價券 100 元）

```python
# === 1. 原始資料（測試一）===
event_name = "  WEEKEND DINNER  "
meal_amount = " NT$1,200 "
people = "4"

service_rate = 0.10
discount = 100
budget_per_person = 300

# === 2. 整理聚餐名稱：去除頭尾空白、轉小寫 ===
clean_event_name = event_name.strip().lower()

# === 3. 整理餐費：移除 NT$、逗號、空白，再轉成數字（保留原始 meal_amount）===
clean_meal_text = meal_amount.replace("NT$", "").replace(",", "").strip()
meal_value = float(clean_meal_text)

# 文字人數 "4" 轉成整數 4，才能做除法分攤
people_count = int(people)

# === 4. 分步計算費用 ===
service_fee = meal_value * service_rate           # 服務費：以「原始餐費」計算
total_payable = meal_value + service_fee - discount  # 應付總額：先加服務費，再扣折價券
per_person = total_payable / people_count          # 每人費用

# === 5. 判斷是否超過預算（布林值）===
over_budget = per_person > budget_per_person

# === 6. 輸出費用摘要 ===
print("===== 聚餐費用摘要 =====")
print(f"聚餐名稱：{clean_event_name}")
print(f"原始餐費：{meal_value:.2f} 元")
print(f"服務費（{service_rate:.0%}）：{service_fee:.2f} 元")
print(f"折價券：{discount:.2f} 元")
print(f"應付總額：{total_payable:.2f} 元")
print(f"分攤人數：{people_count} 人")
print(f"每人費用：{per_person:.2f} 元")
print(f"每人預算：{budget_per_person:.2f} 元")
print(f"是否超過預算：{over_budget}")
```

```python
# 檢查型別轉換前後的差異
print(type(meal_amount), type(meal_value))
print(type(people), type(people_count))
print(type(over_budget))
```
